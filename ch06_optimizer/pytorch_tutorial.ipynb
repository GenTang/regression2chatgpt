```python
import torch

# Create tensors
## Create tensors with convenience functions
zeros = torch.zeros(2, 3)
print(zeros)

ones = torch.ones(2, 3)
print(ones)

torch.manual_seed(1024)
random = torch.rand(3, 4)
print(random)
```

```python
# Create tensors
## Create from Python objects
data = [[2, 3, 4], [1, 0, 1]]
t_data = torch.tensor(data)
print(t_data)

## Create from NumPy objects
import numpy as np

n_data = np.array(data)
tn_data = torch.from_numpy(n_data)
print(tn_data)

## NumPy bridge: changes to a NumPy object propagate to the tensor
n_data += 1
torch.all(torch.from_numpy(n_data) == tn_data)
```

```python
# Change tensor dimensions
## Increase or decrease the number of dimensions
a = torch.rand(3, 4)
print(a.shape)
## Increase dimensions
b = a.unsqueeze(0)
print(b.shape)
## Decrease dimensions
c = b.squeeze(0)
print(c.shape)
## Same data, different dimensions
print(torch.all(c.eq(b)))
print(c.shape == b.shape)
```

```python
# Reshape tensors
data = torch.tensor(range(0, 10))
print(data, data.shape)
view1 = data.view(2, 5)
print(view1)
transpose1 = view1.T
print(transpose1)
```

```python
## Non-contiguous objects cannot use the view operation
print(view1.is_contiguous(), transpose1.is_contiguous())
view2 = transpose1.view(1, 10)
```

```python
# Element-wise operations
twos = torch.ones(2, 2) * 2
print(twos)

powers = twos ** torch.tensor([[1, 2], [3, 4]])
print(powers)
```

```python
## Tensor broadcasting
a = torch.tensor(range(1, 7)).view(2, 3)
b = torch.tensor(range(1, 4)).view(   3)
print(a)
print(b)
print(a * b)

## A more complex broadcasting example
a =     torch.ones(4, 1, 3, 2)
b = a * torch.rand(   5, 1, 2)
print(b.shape)
```

```python
# Matrix operations
mat1 = torch.randn(3, 4)
mat2 = torch.randn(4, 5)
re = mat1 @ mat2
print(re.shape)
## Broadcasting in matrix operations
mat1 = torch.randn(5, 1, 3, 4)
mat2 = torch.randn(   8, 4, 5)
re = mat1 @ mat2
print(re.shape)
```

```python
# Vector operations
# Vector and vector
vec1 = torch.randn(3)
vec2 = torch.randn(3)
print((vec1 @ vec2).shape)
# Matrix and vector
mat = torch.randn(3, 4)
vec = torch.randn(4)
print((mat @ vec).shape)
# Tensor and vector
mat = torch.randn(10, 3, 4)
vec = torch.randn(4)
print((mat @ vec).shape)
```
