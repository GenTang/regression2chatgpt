```python
import torch

# Set a fixed random seed for reproducible results
torch.manual_seed(1024)
# Generate the training data
x_origin = torch.linspace(100, 300, 200)
# Normalize X; otherwise, gradient descent can easily become unstable
x = (x_origin - torch.mean(x_origin)) / torch.std(x_origin)
epsilon = torch.randn(x.shape)
y = 10 * x + 5 + epsilon
```

```python
# Create the model
model = Linear()
# Define the batch size
batch_size = 20
learning_rate = 0.1

for t in range(20):
    # Select the current batch for model training
    ix = (t * batch_size) % len(x)
    xx = x[ix: ix + batch_size]
    yy = y[ix: ix + batch_size]
    # Calculate the loss on the current batch
    loss = mse([model.error(_x, _y) for _x, _y in zip(xx, yy)])
    # Calculate the gradient of the loss
    loss.backward()
    # Iteratively update the model parameter estimates
    model.a -= learning_rate * model.a.grad
    model.b -= learning_rate * model.b.grad
    # Clear gradients after use
    model.a.grad = 0.0
    model.b.grad = 0.0
    print(f'Step {t + 1}, Result: {model.string()}')
```
