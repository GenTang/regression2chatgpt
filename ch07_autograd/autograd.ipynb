```python
from utils import Scalar, draw_graph
```

```python
# A simple computation graph
a = Scalar(1.0, label='a')
b = Scalar(2.0, label='b')
c = a + b
draw_graph(c)
```

```python
# A slightly more complex computation graph
a = Scalar(1.0, label='a')
b = Scalar(2.0, label='b')
c = Scalar(4.0, label='c')
d = a + b
e = a * c
f = d * e
backward_process = f.backward(draw_graph)
draw_graph(f)
```

```python
draw_graph(f, 'backward')
```

```python
# Display the backpropagation process (a window may open)
for index, pic in enumerate(backward_process):
    pic.view(str(index))
```
