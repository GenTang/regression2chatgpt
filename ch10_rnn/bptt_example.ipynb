```python
import os
import sys
sys.path.append(os.path.abspath(os.path.join('..')))
from ch07_autograd.utils import Scalar, draw_graph
```

```python
# Define the computation graph
w = Scalar(0.3, label='w')
wh = Scalar(0.5, label='wh')
x1 = Scalar(1., label='x1', requires_grad=False)
x2 = Scalar(1., label='x2', requires_grad=False)
x3 = Scalar(1., label='x3', requires_grad=False)
# Treat h1, h2, and h3 as hidden states
# They are the hidden states at the first, second, and third steps, respectively
h1 = w * x1
h2 = w * x2 + wh * h1
h3 = w * x3 + wh * h2
```

```python
# Backpropagate through the hidden state at the first step
h1.backward()
draw_graph(h1, 'backward')
```

```python
# Backpropagate through the hidden state at the second step
h2.backward()
draw_graph(h2, 'backward')
```

```python
# Backpropagate through the hidden state at the third step
h3.backward()
draw_graph(h3, 'backward')
```
